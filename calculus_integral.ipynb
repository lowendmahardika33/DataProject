```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  

def C(x, y):
    return x**2 + 2*(y**2) - 4*x - 6*y + x*y + 100

A = np.array([[2.0, 1.0],
              [1.0, 4.0]])
b = np.array([4.0, 6.0])
x_star, y_star = np.linalg.solve(A, b)
z_star = C(x_star, y_star)

xx = np.linspace(-1.0, 4.0, 1000)
yy = np.linspace(-1.0, 4.0, 1000)
X, Y = np.meshgrid(xx, yy)
Z = C(X, Y)

fig = plt.figure(figsize=(7, 5))
ax = fig.add_subplot(111, projection="3d")
ax.plot_surface(X, Y, Z, linewidth=0, antialiased=True)
ax.scatter(x_star, y_star, z_star, s=60)
ax.set_title("3D surface of C(x, y) with minimum marked")
ax.set_xlabel("x (raw material)")
ax.set_ylabel("y (labor hours)")
ax.set_zlabel("C(x, y)")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  

def f(x, y):
    return x**2 - y**2

H = np.array([[2.0, 0.0],
              [0.0, -2.0]])
detH = np.linalg.det(H)
print("Hessian:\n", H)
print("det(H) =", detH, "< 0 -> bentuk pelana (saddle)")

xx = np.linspace(-3.0, 3.0, 200)
yy = np.linspace(-3.0, 3.0, 200)
X, Y = np.meshgrid(xx, yy)
Z = f(X, Y)

plt.figure(figsize=(6, 5))
cs = plt.contour(X, Y, Z, levels=25)
plt.clabel(cs, inline=True, fontsize=8)
plt.scatter([0], [0], s=60, marker="o")  # titik pelana
plt.title("Contour f(x, y) = x^2 - y^2 (det Hessian < 0, pelana)")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()

fig = plt.figure(figsize=(7, 5))
ax = fig.add_subplot(111, projection="3d")
ax.plot_surface(X, Y, Z, linewidth=0, antialiased=True)
ax.scatter(0, 0, 0, s=60) 
ax.set_title("3D surface f(x, y) = x^2 - y^2 (pelana)")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("f(x, y)")
plt.show()
```
